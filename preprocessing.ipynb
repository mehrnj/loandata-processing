```python
df = df.drop("credit_score" , axis=1)
df
```

```python
def gender(s):
    if s == "male":
        return 1
    else:
        return 0
df["person_gender"] = df["person_gender"].apply(gender)
df
```

```python
male = df[df["person_gender"] == 1]
female = df[df["person_gender"] == 0]
```

```python
import matplotlib.pyplot as plt
plt.pie([len(male) , len(female)] , labels=["male" , "female"])
plt.show()
```
